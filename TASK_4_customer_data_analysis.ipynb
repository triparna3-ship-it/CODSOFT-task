```python
df["Customer_ID"].unique()             #list 0f unique id
df["Customer_ID"].nunique()            #number of unique id / Total_customer
```

```python
top_customer = df.groupby("Customer_ID")["Revenue"].sum()                          #Top 10 Customers by Revenue

top_customer = top_customer.sort_values(ascending=False).head(10)

print(top_customer)
```

```python
plt.figure(figsize=(10,5))                                #Top 10 Customers

plt.bar(top_customer.index, top_customer.values)

plt.title("Top 10 Customers by Revenue")
plt.xlabel("Customer ID")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.show()
```

```python
customer_revenue = df.groupby("Customer_ID")["Revenue"].sum()                #Most valuable customers by total revenue

customer_revenue = customer_revenue.sort_values(ascending=False)

print(customer_revenue.head(10))
```

```python
state_revenue = df.groupby("State")["Revenue"].sum()                   #Most valuable customer groups by State

state_revenue = state_revenue.sort_values(ascending=False)

print(state_revenue)
```

```python
city_revenue = df.groupby("City")["Revenue"].sum()           #Most valuable customer groups by City

city_revenue = city_revenue.sort_values(ascending=False)

print(city_revenue.head(10))
```

```python
category_revenue = df.groupby("Category")["Revenue"].sum()                   #Most valuable groups by Category

category_revenue = category_revenue.sort_values(ascending=False)

print(category_revenue)
```

```python
customer_revenue = df.groupby("Customer_ID")["Revenue"].sum()                      #Find high-value customers

high = customer_revenue.quantile(0.75)
low = customer_revenue.quantile(0.25)

def segment(revenue):
    if revenue >= high:
        return "High Value"
    elif revenue >= low:
        return "Medium Value"
    else:
        return "Low Value"

customer_segment = customer_revenue.apply(segment)

print(customer_segment.value_counts())
```

```python
state=df["State"].value_counts()
plt.figure(figsize=(7,7))

plt.pie(
    state,
    labels=state.index,
    autopct="%1.1f%%"
)

plt.title("Customers by State")

plt.show()
```

```python
category = df.groupby("Category")["Quantity"].sum()           #Buying Pattern (Most Purchased Category)

print(category)
```

```python
channel = df["Sales_Channel"].value_counts()                               #Sales Channel Preference

print(channel)
```

```python
payment = df["Payment_Method"].value_counts()                               #Payment Method

print(payment)
```

```python
plt.figure(figsize=(8,5))

plt.bar(payment.index, payment.values)

plt.title("Payment Method")
plt.xlabel("Payment")
plt.ylabel("Customers")

plt.show()
```

```python
customer_value = df.groupby("Customer_ID")["Revenue"].sum()                          #High-Value Customers

average = customer_value.mean()

high_value = customer_value[customer_value > average]

print(high_value)
```

```python
high_value = high_value.sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))

plt.bar(high_value.index, high_value.values)

plt.title("Top High-Value Customers")
plt.xlabel("Customer ID")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.show()
```

```python
print("Total Customers:", df["Customer_ID"].nunique())                                 #Business Insights

print("Highest Revenue:",
      df.groupby("Customer_ID")["Revenue"].sum().max())

print("Most Popular Category:",
      df.groupby("Category")["Quantity"].sum().idxmax())

print("Most Used Payment Method:",
      df["Payment_Method"].mode()[0])

print("Most Preferred Sales Channel:",
      df["Sales_Channel"].mode()[0])
```
