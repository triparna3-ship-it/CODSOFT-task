```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"C:\Users\sutra\Downloads\Advanced_Retail_Sales_Dataset_Cleaned_1030rows.csv")
```

```python
category = df.groupby("Category")["Revenue"].sum()                             #RevenuevsCategory_BAR_CHART

plt.figure(figsize=(8,5))

plt.bar(category.index, category.values)

plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.show()
```

```python
#df["Sale_Date"] = pd.to_datetime(df["Sale_Date"])                          #LINE_Chart
df["Sale_Date"] = pd.to_datetime(df["Sale_Date"], format="%d/%m/%Y")
monthly = df.groupby(df["Sale_Date"].dt.month)["Revenue"].sum()

plt.figure(figsize=(8,5))

plt.plot(monthly.index, monthly.values, marker="o")

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.grid()

plt.show()
```

```python
sales = df["Sales_Channel"].value_counts()                 #PIE_CHART

plt.figure(figsize=(6,6))

plt.pie(
    sales,
    labels=sales.index,
    autopct="%1.1f%%"
)

plt.title("Sales Channel Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))                               #HISTOGRAM

plt.hist(df["Revenue"], bins=20)

plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))                              #SCATTER

plt.scatter(df["Revenue"], df["Profit"])

plt.title("Revenue vs Profit")
plt.xlabel("Revenue")
plt.ylabel("Profit")

plt.show()
```

```python
city = df.groupby("City")["Revenue"].sum()          

city = city.sort_values(ascending=False).head(10)

plt.figure(figsize=(9,5))

plt.bar(city.index, city.values)

plt.title("Top 10 Cities by Revenue")
plt.xlabel("City")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.show()
```
